```python
import pandas as pd
import requests
import time

from tqdm import tqdm
tqdm.pandas()
```

```python
def get_news_for_a_period(start_date: str, end_date: str, countries: list = ['US', 'FR', 'GB', 'EU', 'AU', 'DE', 'JP', 'CN']):
    start_date += 'T00:00:00'
    end_date += 'T00:00:00'

    url = 'https://economic-calendar.tradingview.com/events'
    headers = {
        'Origin': 'https://in.tradingview.com'
    }
    payload = {
        'from': start_date + '.000Z',
        'to': end_date + '.000Z',
        'countries': ','.join(countries)
    }
    data = requests.get(url, headers=headers, params=payload).json()
    data = pd.DataFrame(data['result'])
    return data
```

```python
# freq='MS' означает Month Start (начало месяца)
dates = pd.date_range(start='2024-01-01', periods=13, freq='MS').strftime('%Y-%m-%d').tolist()
```

```python
print(dates)
```

```python
calendar = []
skipped_dates = []
pb = tqdm(range(1, len(dates)))
for i in pb:
    pb.set_description(f'Fetching {start_date} - {end_date}')
    start_date = dates[i - 1]
    end_date = dates[i]
    calendar_chunk = get_news_for_a_period(start_date, end_date)
    if calendar_chunk.empty:
        pb.set_postfix_str('Empty chunk. Sleeping... (5 sec) and try again')
        time.sleep(5)
        calendar_chunk = get_news_for_a_period(start_date, end_date)
        if calendar_chunk.empty:
            pb.set_postfix_str('Calendar chunk is empty again. Skipping...')
            skipped_dates.append((start_date, end_date))
            time.sleep(3)
    calendar.append(calendar_chunk)
    pb.set_postfix_str('Sleeping... (3 sec)')
    time.sleep(3)

calendar = pd.concat(calendar)
```

```python
def get_news_for_a_year(date_list: list):
    calendar = []
    skipped_dates = []

    pb = tqdm(range(1, len(date_list)))
    for i in pb:
        start_date = date_list[i - 1]
        end_date = date_list[i]

        pb.set_description(f'Fetching {start_date} - {end_date}')

        calendar_chunk = get_news_for_a_period(start_date, end_date)
        
        if calendar_chunk.empty:
            pb.set_postfix_str('Empty chunk. Sleeping... (5 sec) and try again')
            time.sleep(5)
            calendar_chunk = get_news_for_a_period(start_date, end_date)
            if calendar_chunk.empty:
                pb.set_postfix_str('Calendar chunk is empty again. Skipping...')
                skipped_dates.append((start_date, end_date))
                time.sleep(3)
        
        calendar.append(calendar_chunk)
        pb.set_postfix_str('Sleeping... (3 sec)')
        time.sleep(3)

    calendar = pd.concat(calendar)
    
    if skipped_dates:
        print(skipped_dates)

    return calendar
```

```python
for year in tqdm([2013, 2014, 2015, 2016, 2017, 2018, 2019]):
    print(f'======== {year} ========')
    dates = pd.date_range(start=f'{year}-01-01', periods=13, freq='MS').strftime('%Y-%m-%d').tolist()
    calendar = get_news_for_a_year(dates)
    calendar = calendar[calendar['date'] < dates[-1]].drop_duplicates()
    calendar.to_csv(f'../../../dataset/tradingview_economy_calendar/calendar{year}.csv', index=False)
    time.sleep(3)
```
